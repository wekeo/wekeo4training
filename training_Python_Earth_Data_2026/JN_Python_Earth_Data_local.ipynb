# 1. Introduction

[Go back to the "Table of contents"](#Table-of-contents)

Vegetation plays an important role in the Earth system. It supports biodiversity, stores carbon, influences the water cycle, and responds to changes in rainfall, temperature, and land management. Monitoring vegetation over time helps us understand how ecosystems change and how they respond to environmental pressures.

One way to monitor vegetation from space is through the **Normalized Difference Vegetation Index (NDVI)**. Calculated from satellite observations of reflected light, NDVI provides an indicator of vegetation greenness. Higher NDVI values generally indicate greener, more actively growing vegetation, while lower values may indicate sparse or stressed vegetation. NDVI changes naturally throughout the year as plants grow and dry out, but it can also vary between years in response to weather conditions and other environmental factors.

In this tutorial, we focus on the **Tabernas area in Almería, southeastern Spain**, a semi-arid landscape where vegetation is naturally sparse, and water availability is an important environmental constraint. This makes it an interesting case study for exploring seasonal vegetation cycles and differences in greenness between years.

Using satellite NDVI data from the Copernicus Land Monitoring Service, we will work with Python to access and explore the data, create maps, plot time series, compare seasonal patterns, and calculate NDVI anomalies. These analyses will help us investigate when and where vegetation is greener or less green than a reference period.

By the end of the tutorial, you will have practised essential Python skills for working with Earth observation data and learned how to use NDVI to explore vegetation variability. We will also consider the limitations of this indicator: changes in NDVI can reveal patterns in vegetation greenness, but they do not directly measure drought or explain the causes of environmental change.

Once these credentials are entered, the `hda` client will automatically read them from the file when you use it. Let's create our credentials file, if it doesn't already exist.

```python
# load credentials
wekeo_credentials_file = Path(Path.home() / ".hdarc")


def get_credentials():
    user = input("Enter your WEkEO user name: ").strip()
    password = getpass.getpass("Enter your WEkEO password: ").strip()
    return user, password


def save_credentials(user, password):
    os.makedirs(os.path.dirname(wekeo_credentials_file), exist_ok=True)
    with open(wekeo_credentials_file, "w", encoding="utf-8") as f:
        f.write(f"user: {user}\npassword: {password}")


# Try loading saved credentials
if not wekeo_credentials_file.is_file():
    user, password = get_credentials()
    save_credentials(user, password)
```

```python
# Try to get a token, retrying if the credentials are invalid
while True:
    try:
        token = hda.Client().token
        print("WEkEO token created.")
        break  # Exit loop if successful
    except hda.api.ConfigurationError as e:
        print("Invalid username or password. Please try again.")
        print(e)
        user, password = get_credentials()
        save_credentials(user, password)
```

## 3.3. Downloading more NDVI data

[Go back to the "Table of contents"](#Table-of-contents)

For further analysis, we need more NDVI data. We will now download NDVI for 21st of each month for the period of 2015-2025. As you might have noticed, the previously downloaded product weighs 2.3 GB; this is due to its spatial resolution. The product covers the entire globe since spatial subsetting does not work for this NDVI product. We do not need coverage like that, so we will subset each downloaded file to our area of interest (AOI), save it as a new file, and delete the original file. This will not only allow us to work solely with the targeted zone, but also will keep our JupyterHub workspace storage under the limit (20 GB).

Let's start downloading more data!

First, we create a 'general' query. As you can see, a query can be as well given as a dictionary.

```python
query = {'dataset_id': 'EO:CLMS:DAT:NDVI_300M_10_DAILY_V3',
 'startdate': '2016-09-15T00:00:00.000Z',
 'enddate': '2016-09-25T23:59:59.999Z',
 'fileFormat': 'nc',
 'sortOrder': 'ASC',
 'itemsPerPage': 200,
 'startIndex': 0}
```

Next, we create a function that allows us to download files, crop them, save them as new files and delete the original ones. This function might appear overwhelming at first (especially for beginner users), but don't worry! We already learned how to search a query and download data matching it, so we repeat the same steps here. After that, we integrate data subsetting. Now, we execute the next cells in order to obtain the data, but we will explain how to subset a NetCDF file in the next section.

```python
# Download file, crop to AOI, save as new file, delete original file
def download_crop_ndvi(year, month, query, output_path,
                        target_day=21,
                        lon_min=-3.5, lon_max=-1.5,
                        lat_min=36.5, lat_max=37.5):

    # Create output folder if it does not already exist
    os.makedirs(output_path, exist_ok=True)

    # Name for output file
    output_file = os.path.join(output_path, f"NDVI_{target_day:02d}-{month:02d}-{year}.nc")
    
    # Search for the target date
    q = deepcopy(query)
    q["startdate"] = f"{year}-{month:02d}-{target_day:02d}T00:00:00.000Z"
    q["enddate"] = f"{year}-{month:02d}-{target_day:02d}T23:59:59.999Z"

    print(f"\nProcessing {year}-{month:02d}-{target_day:02d}")

    # Download the file
    results = c.search(q)
    results.download(output_path)

    # Locate the downloaded file
    pattern = os.path.join(output_path, f"c_gls_NDVI300_{year}{month:02d}{target_day:02d}*.nc")
    source = glob.glob(pattern)[0]

    # Open and crop the dataset
    with xr.open_dataset(source) as ds:
        subset = ds.sel(lon=slice(lon_min, lon_max), lat=slice(lat_max, lat_min))
        subset.load()

    # Save the cropped dataset
    subset.to_netcdf(output_file)

    # Delete the original global file
    os.remove(source)

    print(f"Saved cropped file: {output_file}")
```

```python
# Download monthly NDVI data from 2015 to 2025
for year in range(2015, 2026):
    for month in range(1, 13):
        download_crop_ndvi(year, month, query, output_path)
```

## 4.2 Subsetting a dataset

[Go back to the "Table of contents"](#Table-of-contents)

From the dataset details above, you can see that the downloaded file covers the whole globe (look at **lat** and **lon** coordinates). Since we are interested only in a small region in southern Spain, we will now use an essential feature of xarray: the `sel` function. It allows us to extract subsets of a dataset along a chosen dimension. In this section, we will use it to select only the NDVI data inside our area of interest (AOI). 

```python
# Define bounding box
lon_min, lon_max = -3.5, -1.5
lat_min, lat_max = 36.5, 37.5
```

From the previously displayed dataset, we could have noticed that latitude dimension is descending. The reason is that satellite datasets commonly store latitude coordinates from north to south (from larger to smaller latitude values). This is why in the next cell, our latitude subset is reversed compared with our longitude subset.

```python
# Select the zone of interest
ds_subset_2015 = ds_ndvi_2015.sel(
    lon=slice(lon_min, lon_max),
    lat=slice(lat_max, lat_min))

ds_subset_2015
```

We can see that now **lat** and **lon** coordinates correspond to a smaller zone.

Let's save this new dataset as a new file! By doing so, we can then delete the original file and save storage space in our workspace.

```python
ds_subset_2015.to_netcdf("data/NDVI_21-09-2015.nc")
```

## 4.3 Preparing a dataset

[Go back to the "Table of contents"](#Table-of-contents)

To further prepare our dataset, we will first have a look at its attributes:

`ndvi_2015.attrs` contains descriptive information about the content — for example, what NDVI means or its units.

`ndvi_2015.encoding` contains technical instructions about how the content is stored — for example, how missing values are represented or how the data is compressed.

```python
# Keep only NDVI variable
ds_subset_2015 = xr.open_dataset("data/NDVI_21-09-2015.nc")
ndvi_2015 = ds_subset_2015["NDVI"].squeeze()

# Print attributes
print("Attributes:", ndvi_2015.attrs)
print("\nEncoding:", ndvi_2015.encoding)
```

From those details, we see that NDVI values vary between 0 and 250, with 252-255 being flag values used to mark invalid cells. We will use this information to prepare a clean dataset with masked invalid points.

Encoding details tell us that a scale factor of 0.004 and an offset of -0.08 were applied to the NDVI values to obtain a traditional scale from 0 to 1. Hence, to obtain the flag values to mask, we can use the next formula:

```python
print("Flag values to mask:", [252 * 0.004 - 0.08, 253 * 0.004 - 0.08, 254 * 0.004 - 0.08, 255 * 0.004 - 0.08])
```

Now we can apply a mask based on these values.

```python
# Mask water and other invalid NDVI values
ndvi_clean_2015 = ndvi_2015.where(ndvi_2015 < 0.92)
ndvi_clean_2015
```

Now, let's arrange the previous steps in a callable function:

```python
def clean_ndvi(path):
    ds = xr.open_dataset(path)
    ndvi = ds["NDVI"].squeeze()
    ndvi_clean = ndvi.where(ndvi < 0.92)
    return ndvi_clean
```

## 4.4 Plotting NDVI map 

[Go back to the "Table of contents"](#Table-of-contents)

In this section, we'll use **matplotlib** and **cartopy** to visualise NDVI for September 21st for each year of 2015-2025.
Notably, we are going to plot the NetCDF data using the `pcolormesh` function.

We start by creating a **Matplotlib figure** and a **map axis** using a geographic projection. The `PlateCarree` projection is well suited for data defined on a regular longitude–latitude grid, such as the data we downloaded.

We then add basic geographic context to the map, including coastlines, national borders, and gridlines with longitude and latitude labels.
Finally, we restrict the map to our region of interest by setting the geographic extent to the selected bounding box.

```python
# Create the figure and map axis
fig = plt.figure(figsize=(10, 8))
ax = plt.axes(projection=ccrs.PlateCarree())

# Add basic geographic features
ax.add_feature(cfeature.COASTLINE)
ax.add_feature(cfeature.BORDERS, linestyle=":")
ax.gridlines(draw_labels=True)
ax.set_extent([lon_min, lon_max, lat_min, lat_max])
```

To avoid repeating the same code, we wrap these steps into a reusable function.
This function adds standard geographic features to any map axis and optionally sets the map extent to a chosen bounding box.

```python
# Let's save this in a function we will reuse later on.
def basic_geo_features(ax, lon_min=None, lon_max=None, lat_min=None, lat_max=None):

    # Add geographic features
    ax.add_feature(cfeature.COASTLINE)  # coastlines
    ax.add_feature(cfeature.BORDERS, linestyle=":")  # country borders (dotted)
    ax.gridlines(draw_labels=True)  # gridlines with labels
    ax.set_extent([lon_min, lon_max, lat_min, lat_max])
```

We now plot **NDVI** on the map using `pcolormesh`, which displays the data as a colored grid. Fixed colour limits are applied.

Finally, we add a colorbar to indicate the NDVI values associated with each colour on the map.
A clear title is added to describe both the variable and the time period represented in the figure.

```python
# Plot NDVI (lon, lat, NDVI field)
im = ax.pcolormesh(
    ndvi_clean_2015.lon,  # longitudes (x)
    ndvi_clean_2015.lat,  # latitudes  (y)
    ndvi_clean_2015,  # NDVI field (z)
    cmap="YlGn",  # colormap
    vmin=0,
    vmax=1,  # set fixed colour limits for comparability
)

# Add a colorbar linked to the plotted image, label it and shrink it to fit the layout
cbar = plt.colorbar(im, ax=ax, label="NDVI", shrink=0.5, pad=0.1)

# Add a title (newline used to wrap subtitle)
ax.set_title("NDVI on September 21, 2015")

fig
```

You’ve just seen how to build a map step by step. We can combine all these steps in a single plotting function:

```python
def plot_ndvi(ndvi_clean, title):
    # Create the figure and map axis
    fig = plt.figure(figsize=(10, 8))
    ax = plt.axes(projection=ccrs.PlateCarree())
    
    # Add basic geographic features
    basic_geo_features(ax, lon_min, lon_max, lat_min, lat_max)
    
    # Plot the NDVI (lon, lat, NDVI field)
    im = ax.pcolormesh(
        ndvi_clean.lon,  # longitudes (x)
        ndvi_clean.lat,  # latitudes  (y)
        ndvi_clean,  # NDVI field (z)
        cmap="YlGn",  # colormap
        vmin=0,
        vmax=1,  # set fixed colour limits for comparability
    )
    
    # Add a colorbar linked to the plotted image, label it and shrink it to fit the layout
    cbar = plt.colorbar(im, ax=ax, label="NDVI", shrink=0.5, pad=0.1)
    
    # Add a title
    ax.set_title(title)

    plt.show()
```

# 5. NDVI time series

[Go back to the "Table of contents"](#Table-of-contents)

Time series are one of the most common ways to study variables because they allow us to track changes over time at a glance. By averaging over space, we can summarise the evolution of a variable - like NDVI - across an entire region, reducing local variability and focusing on broader trends or seasonal cycles.

## 5.1 From xarray dataset to time series

[Go back to the "Table of contents"](#Table-of-contents)

In this section, we will compute a spatially averaged time series of NDVI by taking the mean over all longitudes and latitudes in our datasets (remember the `mean` function?). This will provide a clear view of how the vegetation cover has evolved over the years and will serve as the basis for further analyses, such as detecting trends, anomalies, or extreme events.

First, let's calculate the NDVI time series for the month of September for 10 years, from 2015 to 2025.

```python
years = range(2015, 2026)

# Create an empty list to accumulate the results
results = []

# Loop through the files
for year in years:
    # We want each September file
    path = f"data/NDVI_21-09-{year}.nc"

    # Prepare the file
    ndvi = clean_ndvi(path)

    # Calculate spatially averaged time series
    mean_ndvi = ndvi.mean(dim=["lon", "lat"], skipna=True).item()

    # Populate the list with results
    results.append({"year": year, "mean_ndvi": mean_ndvi})

# Turn results in dataframe
df_september = pd.DataFrame(results)

# Print it
print(df_september.to_string(index=False))
```

We now have our time series for September 2015-2025 in the year-mean NDVI form. 
Let's plot it!

```python
# Plot the calculated time series
fig, ax = plt.subplots(figsize=(10, 5))

ax.plot(
    df_september["year"],
    df_september["mean_ndvi"],
    marker="o",
    linewidth=2
)

ax.set_xlabel("Year")
ax.set_ylabel("Mean NDVI")
ax.set_title("September NDVI evolution, 2015–2025")
ax.set_xticks(df_september["year"]) # Set ticks for each year
ax.grid(True, alpha=0.3) # Add grid

plt.tight_layout()
plt.show()
```

Looking at the September time series from 2015 to 2025, we don't see a clear upward or downward trend in mean NDVI. Vegetation greenness varies from year to year: for example, 2023 has the **highest mean NDVI**, while **2016 has the lowest**. This suggests that vegetation conditions fluctuate over time, but these September observations alone don't show a consistent long-term change.

But what if we look at the whole year, rather than just September? Let's plot NDVI for every month from 2015 to 2025. This will help us explore seasonal changes in vegetation and see whether any patterns become clearer over time.

First, we calculate the new full time series.

```python
data_dir = Path("data")
records = []

# Loop through all files
for path in sorted(data_dir.glob("NDVI_*.nc")):
    # Extract date from file name
    date = pd.to_datetime(path.stem.replace("NDVI_", ""), format="%d-%m-%Y")

    # Prepare dataset
    ndvi = clean_ndvi(str(path))
    
    # Calculate mean
    mean_ndvi = ndvi.mean(dim=["lon", "lat"], skipna=True).item()

    records.append({"date": date, "mean_ndvi": mean_ndvi})

df_monthly = pd.DataFrame(records).sort_values("date").reset_index(drop=True)
```

And plot it.

```python
fig, ax = plt.subplots(figsize=(12, 5))
ax.plot(df_monthly["date"], df_monthly["mean_ndvi"], marker="o")

ax.set_xlabel("Date")
ax.set_ylabel("Area-weighted mean NDVI")
ax.set_title("Vegetation dynamics through time")
ax.xaxis.set_major_locator(matplotlib.dates.MonthLocator(interval=2)) # Set ticks for each second month
ax.tick_params(axis="x", labelrotation=90) 
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

The time series shows a clear seasonal pattern in vegetation greenness. NDVI generally increases during the greener parts of the year and decreases during the drier summer months. The height of these seasonal peaks varies from year to year, meaning that some years have greener vegetation than others.

However, there is no obvious long-term increase or decrease in greenness across the whole period. The data show seasonal changes and differences between years, but we cannot conclude from this plot alone that the region is becoming greener or browner over time.

# 6. NDVI anomalies

[Go back to the "Table of contents"](#Table-of-contents)

So far, we have explored how vegetation greenness changes throughout the year and how it varies between years. However, NDVI naturally rises and falls with the seasons, making it difficult to identify unusual changes just by looking at the original time series.

To address this, we calculate a seasonal reference cycle: the average NDVI for each calendar month across all available years. For example, the reference for January is the average of all January observations.

We then calculate an NDVI anomaly for each month by subtracting its seasonal reference from the observed NDVI:

$$\text{Anomaly} = \text{Observed NDVI} - \text{Reference NDVI} $$
                                                
- **Positive anomaly**: vegetation is greener than the usual value for that calendar month.
- **Negative anomaly**: vegetation is less green than the usual value for that calendar month.
- **Anomaly near zero**: vegetation greenness is close to the reference value.
  
This helps us distinguish unusual greenness from the normal seasonal cycle.


## 6.1 Calculate and plot reference seasonal cycle

[Go back to the "Table of contents"](#Table-of-contents)


First, we calculate a seasonal reference cycle.

```python
# Calendar month for each observation
df = df_monthly
df["month"] = df_monthly["date"].dt.month

# Reference mean for each calendar month, across all years
reference_cycle = df.groupby("month")["mean_ndvi"].mean().rename("reference_ndvi")

# Add the reference and calculate anomalies
df = df.join(reference_cycle, on="month")
```

The previous cell extracts the calendar month from each date and calculates the average NDVI for each month across all years. The result contains 12 reference values, one for each month. Then, the monthly reference is added to each observation. Each observation is compared with the reference for the same calendar month, rather than with one average for the whole year.

```python
# Let's have a look at the new dataset
df
```

Now, we move to plotting.

```python
# Plot the observed time series and reference seasonal cycle
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(df["date"], df["mean_ndvi"],
    label="Observed NDVI", marker=".", linewidth=1)

ax.plot(df["date"], df["reference_ndvi"],
    label="Calendar-month reference", linewidth=2)

ax.set_xlabel("Date")
ax.set_ylabel("Mean NDVI")
ax.set_title("Observed NDVI compared with the reference seasonal cycle")
ax.legend()
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

This plot compares the original monthly observations with the seasonal reference.
- The Observed NDVI line shows how vegetation greenness changes over time.
- The Calendar-month reference line shows the average seasonal pattern repeated across the observation period.

This comparison helps us see how individual years differ from the typical seasonal cycle.

## 6.2 Calculate anomalies

[Go back to the "Table of contents"](#Table-of-contents)

Next, we add another column to the previous dataset - anomaly column, and plot it.

```python
# Calculate anomalies
df["anomaly"] = (df["mean_ndvi"] - df["reference_ndvi"])

# Plot the anomalies
fig, ax = plt.subplots(figsize=(12, 4))

ax.axhline(0, color="black", linewidth=1)
ax.bar(df["date"], df["anomaly"], width=20)

ax.set_xlabel("Date")
ax.set_ylabel("NDVI anomaly")
ax.set_title("Monthly NDVI anomalies relative to the reference cycle")
ax.grid(True, axis="y", alpha=0.3)

plt.tight_layout()
plt.show()
```

And now, we see the difference between observed NDVI and the monthly reference.
- Bars above zero indicate greener-than-reference conditions.
- Bars below zero indicate less-green-than-reference conditions.
- Bars close to zero indicate conditions close to the reference.
    
This plot makes it easier to spot periods when vegetation greenness was unusually high or low compared with the same month in other years.

The observed NDVI plot shows both the normal seasonal cycle and variations between years. The anomaly plot removes the average seasonal pattern, making departures from typical monthly greenness easier to identify.

By using `loc` method of `pandas.DataFrame`, we can locate the worst anomaly occurring - the largest negative anomaly.

```python
# Make sure dataset has no NaN values 
valid_anomalies = df.dropna(subset=["anomaly"])

# Locate the biggest negative anomaly
worst = valid_anomalies.loc[valid_anomalies["anomaly"].idxmin()]

print("Date of largest negative anomaly:", worst["date"])
print("Observed NDVI:", worst["mean_ndvi"])
print("Reference NDVI:", worst["reference_ndvi"])
print("Anomaly:", worst["anomaly"])
```

```python
# Select months when NDVI is more than 0.03 below the reference
below = df[df["anomaly"] < -0.03].sort_values("date")

# Variables to store the longest sequence found so far
longest_start, longest_end, longest_length = None, None, 0

# Variables to track the current sequence of consecutive months
start, previous_date, length = None, None, 0

# Go through each date when NDVI is below the threshold
for date in below["date"]:

    # Start a new sequence if this is the first date, or if there is a gap between monthly observations
    if previous_date is None or (date - previous_date).days not in range(28, 32): # days in month can be 28, 29, 30, 31
        start = date
        length = 1
    else:
        # If the observation is in the following month, extend the current sequence
        length += 1

    # Check whether this sequence is the longest so far
    if length > longest_length:
        longest_length = length
        longest_start = start
        longest_end = date
        
    previous_date = date

if longest_length:
    print("Start:", longest_start.date())
    print("End:", longest_end.date())
    print("Consecutive months:", longest_length)
else:
    print("No observations below the selected threshold.")
```

## 6.3 Plot anomaly map

[Go back to the "Table of contents"](#Table-of-contents)

So far, we have used time series to explore how vegetation greenness changes over time. But a regional average cannot tell us where vegetation is becoming greener or less green.

Now, we will use maps to explore the spatial patterns of vegetation greenness across our study area. We will calculate the mean NDVI for each pixel over a year, then compare it with the 2015–2025 climatology — the average annual NDVI for each pixel over the reference period.

The difference between a year's mean NDVI and the climatology is called an annual NDVI anomaly. It shows which areas were greener or less green than the 2015–2025 reference average. By mapping these anomalies, we can identify spatial patterns that were hidden in the regional time series. 

One important distinction: although we call the 2015–2025 average a climatology here, it is a relatively short reference period, not necessarily a formal long-term climate normal. NDVI anomalies also describe differences in vegetation greenness; they do not prove drought or explain the causes of change.

First, we calculate annual means and climatology.

```python
# Load all NDVI files
rasters = {}

# Associate it with its observation date, so we can group observations by year
for path in sorted(Path("data").glob("NDVI_*.nc")):
    date = pd.to_datetime(path.stem.replace("NDVI_", ""), format="%d-%m-%Y")
    rasters[date] = clean_ndvi(str(path))

# Calculate the annual mean NDVI
yearly_means = []

# Average the monthly NDVI values at every pixel
for year in range(2015, 2026):
    yearly_data = [da for date, da in rasters.items() if date.year == year]

    annual_mean = xr.concat(yearly_data, dim="month").mean(dim="month", skipna=True)

    yearly_means.append(annual_mean)

# Calculate the 2015–2025 climatology
climatology = xr.concat(yearly_means, dim="year").mean(dim="year", skipna=True)
```

Let's now plot the climatology map! This map shows the average vegetation greenness at each pixel over the period 2015–2025. It serves as our reference map for comparing annual NDVI maps.

```python
# Plot the climatology
fig, ax = plt.subplots(figsize=(9, 7), subplot_kw={"projection": ccrs.PlateCarree()})

basic_geo_features(ax, lon_min, lon_max, lat_min, lat_max)

im = ax.pcolormesh(climatology.lon, climatology.lat, climatology,
    cmap="YlGn", vmin=0, vmax=1, shading="auto")

fig.colorbar(im, ax=ax, orientation="horizontal", pad=0.12, shrink=0.75, label="NDVI")

ax.set_title(f"NDVI 2015–2025 climatology")
plt.show()
```

Then, we calculate and plot an annual anomaly map.

```python
# Select the year to map
year = 2023
index = year - 2015

# Calculate the anomaly relative to the climatology
anomaly_map = yearly_means[index] - climatology

# Plot the anomaly map
fig, ax = plt.subplots(figsize=(9, 7), subplot_kw={"projection": ccrs.PlateCarree()})

basic_geo_features(ax, lon_min, lon_max, lat_min, lat_max)

im = ax.pcolormesh(anomaly_map.lon, anomaly_map.lat, anomaly_map,
    cmap="RdBu_r", vmin=-0.15, vmax=0.15, shading="auto")

fig.colorbar(im, ax=ax, orientation="horizontal", pad=0.12, shrink=0.75, label="Annual NDVI anomaly")

ax.set_title(f"Annual NDVI anomaly — {year} relative to 2015–2025 climatology")
plt.show()
```

Blue areas have higher annual mean NDVI than the 2015–2025 climatology, while red areas have lower annual mean NDVI. The map reveals where greenness differs from the reference average.

[Go back to the "Table of contents"](#Table-of-contents)

Now, we open our downloaded file and look at its structure.

```python
# Open the downloaded ERA5 NetCDF file
ds_era5 = xr.open_dataset("data/reanalysis-era5-land-monthly-means-4748c6e3478cc0a55b5e1343f33b7752.nc")

# Check the variables, units and dimensions
print(ds_era5)
```

Next, we calculate the average temperature and precipitation over the Tabernas study area. We use the same method of averaging over latitude and longitude as we did for the NDVI analysis.

```python
# Average temperature over the study area, in degrees Celsius
temperature = (ds_era5["t2m"].mean(dim=("latitude", "longitude"), skipna=True)- 273.15)

# Average precipitation over the study area
precipitation = ds_era5["tp"].mean(dim=("latitude", "longitude"), skipna=True)

# Put the variables into a DataFrame
df_climate = pd.DataFrame({
    "date": pd.to_datetime(temperature["valid_time"].values),
    "temperature": temperature.values,
    "precipitation": precipitation.values
})

# Convert precipitation from m to mm
df_climate["precipitation"] = (df_climate["precipitation"] * 1000)

# Show the first 7 lines of the dataframe
df_climate.head(7)
```

Our existing `df` DataFrame contains the NDVI observation dates, regional mean NDVI, reference NDVI, and NDVI anomalies. We'll now combine this DataFrame with the one we just obtained with the ERA5 variables. We'll match records by year and calendar month, rather than requiring the dates to be identical.

```python
# Make a copy of existing NDVI dataset
df_NDVI = df

# Add year and month to both tables
df_NDVI["year"] = df_NDVI["date"].dt.year
df_NDVI["month"] = df_NDVI["date"].dt.month

df_climate["year"] = df_climate["date"].dt.year
df_climate["month"] = df_climate["date"].dt.month

# Combine NDVI and climate observations
df_combined = pd.merge(df_NDVI, df_climate, on=["year", "month"], how="inner")

# Remove rows with missing values
df_combined = df_combined.dropna(subset=["mean_ndvi", "temperature", "precipitation"])

df_combined.head(7)
```

## 7.3 Calculate Pearson correlation
[Go back to the "Table of contents"](#Table-of-contents)

Now, we can calculate Pearson correlations between mean NDVI and each climate variable.

The correlation coefficients range from −1 to +1:
- Close to +1: a strong positive linear relationship.
- Close to −1: a strong negative linear relationship.
- Close to 0: a weak linear relationship.

```python
correlation = df_combined[["mean_ndvi", "temperature", "precipitation"]].corr()
print(correlation)
```

We can now explore the relationship visually with scatter plots. We'll use `np.polyfit()` to calculate a straight line through the points. The line helps show the overall direction even when individual observations are scattered.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# NDVI vs temperature
axes[0].scatter(df_combined["temperature"], df_combined["mean_ndvi"])

x = df_combined["temperature"]
y = df_combined["mean_ndvi"]
slope, intercept = np.polyfit(x, y, 1)
axes[0].plot(x, slope * x + intercept, color="orange")

axes[0].set_xlabel("Temperature (°C)")
axes[0].set_ylabel("Mean NDVI")
axes[0].set_title("NDVI and Temperature")

# NDVI vs precipitation
axes[1].scatter(df_combined["precipitation"], df_combined["mean_ndvi"])

x = df_combined["precipitation"]
y = df_combined["mean_ndvi"]
slope, intercept = np.polyfit(x, y, 1)
axes[1].plot(x, slope * x + intercept, color="orange")

axes[1].set_xlabel("Precipitation")
axes[1].set_ylabel("Mean NDVI")
axes[1].set_title("NDVI and Precipitation")

plt.tight_layout()
plt.show()
```

From the Pearson correlations calculated above and the scatter plots, we can draw the following conclusions:

- **NDVI vs temperature**: \(r=-0.674\)

The points look quite scattered while still showing a negative relationship overall. The negative coefficient means that higher temperatures tend to coincide with lower NDVI in our dataset. In a semi-arid region, this could reflect the seasonal cycle: vegetation may be greener during cooler, wetter periods and browner during the hot, dry summer.

- **NDVI vs precipitation**: \(r=+0.183\)

The weak positive correlation means there is only a slight overall tendency for NDVI to be higher when precipitation is higher. The slight pattern we see may indicate that low precipitation values occur with a range of NDVI values, while high precipitation values are associated with a more restricted range of NDVI.


## 7.4 Plot time series
[Go back to the "Table of contents"](#Table-of-contents)

In this section, we will plot the **monthly time series of NDVI and temperature** over the area of interest. This  will help us explore how vegetation greenness changes alongside temperature over time.

To better understand the relationship between NDVI and temperature, we can plot both variables on the **same figure using dual y-axes**.  

- The **left y-axis** will show NDVI.  
- The **right y-axis** will show temperature (°C).  


```python
# Create figure and primary axis
fig, ax1 = plt.subplots(figsize=(12, 6))

# Plot NDVI on the left y-axis
ax1.plot(df_combined["date_x"], df_combined["mean_ndvi"], color="forestgreen", label="Mean NDVI")
ax1.set_ylabel("Mean NDVI", color="forestgreen")
ax1.tick_params(axis="y", labelcolor="forestgreen")
ax1.grid(True, alpha=0.3)

# Plot temperature on the right y-axis
ax2 = ax1.twinx()
ax2.plot(df_combined["date_x"], df_combined["temperature"], color="darkorange", label="Temperature")
ax2.set_ylabel("Temperature (°C)", color="darkorange")
ax2.tick_params(axis="y", labelcolor="darkorange")

# Add title and format dates
plt.title("Monthly NDVI and Temperature in the Tabernas Area")
fig.autofmt_xdate()
plt.tight_layout()
plt.show()
```

Our time series shows **a clear opposing pattern**: when temperature spikes, NDVI tends to decrease, while cooler periods tend to coincide with higher NDVI. A possible explanation is that high temperatures increase evaporative demand, causing plants to lose water more quickly. In this semi-arid region, where water availability is often limited, this can contribute to vegetation stress and reduced greenness.

However, temperature is not the only factor influencing NDVI. Rainfall, soil moisture, vegetation type, and seasonal changes can also affect the pattern. This visual comparison helps us identify relationships over time, but it cannot establish that temperature directly causes changes in vegetation.
